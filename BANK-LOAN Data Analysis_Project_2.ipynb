### Good Loan Metrics

```python
good_loans = df[df['loan_status'].isin(['Fully Paid','Current'])]

total_loan_applications = df['id'].nunique()

good_loan_applications = good_loans['id'].nunique()
good_loan_funded_amount = good_loans['loan_amount'].sum()
good_loan_recevied = good_loans['total_payment'].sum()

good_loan_funded_amount_millions = good_loan_funded_amount / 1000000
good_loan_recevied_millions = good_loan_recevied / 1000000

good_loan_percentage = (good_loan_applications / total_loan_applications) * 100

print(f"Good Loan Applications: {good_loan_applications}")
print(f"Good Loan Funded Amount: (in Millions): {good_loan_funded_amount_millions:.2f}$")
print(f"Good Loan Total Recived (in Millions): {good_loan_recevied_millions:.2f}$")
print(f"Percentage of Good Loan Applications: {good_loan_percentage:.2f}%")
```

### Bad Loan Metrics

```python
bad_loans = df[df['loan_status'].isin(['Charged Off'])] # Bad Loans

total_loan_applications = df['id'].nunique()

bad_loan_applications = bad_loans['id'].nunique()
bad_loan_funded_amount = bad_loans['loan_amount'].sum()
bad_loan_recevied = bad_loans['total_payment'].sum()

bad_loan_funded_amount_millions = bad_loan_funded_amount / 1000000
bad_loan_recevied_millions = bad_loan_recevied / 1000000

bad_loan_percentage = (bad_loan_applications / total_loan_applications) * 100

print(f"Bad Loan Applications: {bad_loan_applications}")
print(f"Bad Loan Funded Amount: (in Millions): {bad_loan_funded_amount_millions:.2f}$")
print(f"Bad Loan Total Recived (in Millions): {bad_loan_recevied_millions:.2f}$")
print(f"Percentage of Bad Loan Applications: {bad_loan_percentage:.2f}%")
```

### Monthly Trends by Issue Date for Total Funded

```python
monthly_funded = (
    df.sort_values('issue_date')
    .assign(month_name = lambda x: x['issue_date'].dt.strftime('%b'))
    .groupby('month_name', sort=False)['loan_amount']
    .sum()
    .reset_index(name='funded_amount')
)

# convert to millions with 2 decimals
monthly_funded['funded_amount_millions'] = monthly_funded['funded_amount'] / 1_000_000

# total funded amount
total_funded_amt_millions = '$' + str(int(round(monthly_funded['funded_amount_millions'].sum(), 0))) + 'M' 

# plot
fig = px.bar(
    monthly_funded,
    x='month_name',
    y='funded_amount_millions',
    title='<b>Monthly Trends by Total Funded</b>',
    height=600,
    width=1400,
    text=monthly_funded['funded_amount_millions'].round(2))

# update layout
fig.update_layout(
    title_font_size=23,
    title_x=0.5,
    xaxis_title='<b>Months</b>',
    yaxis_title='<b>Total Funded Amount (Millions)</b>',
    font=dict(size=13, family='Arial'),
    margin=dict(l=40, r=40, t=60, b=60),
    annotations=[dict(
        text=f'Funded Amt: {total_funded_amt_millions}',
        x=0.5, y=0.98,
        showarrow=False,
        xref='paper', yref='paper',
        bgcolor='white', bordercolor='black',
        borderpad=4,
        font=dict(size=24, family='Lato', weight='bold')
    )]
)

# update traces
fig.update_traces(
    textposition='outside',
    textfont_size=13,
    textfont_weight='bold',
    hovertemplate='%{x}: %{y:.0f}M',   
    marker=dict(line=dict(color='white', width=3))
)

# update yaxes
fig.update_yaxes(range=[0, monthly_funded['funded_amount_millions'].max() + 10])

fig.show()
```

### Monthly Trends by Issue Date for Total Amount Recevied

```python
import plotly.graph_objects as go

monthly_received = (
    df.sort_values('issue_date')
    .assign(month_name=lambda x: x['issue_date'].dt.strftime('%b'))
    .groupby('month_name', sort=False)['total_payment']
    .sum()
    .div(1_000_000)   # millions
    .reset_index(name='received_amount_millions')
)


labels = monthly_received['received_amount_millions'].round(2).astype(str)

# fig 
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=monthly_received['month_name'],
    y=monthly_received['received_amount_millions'],
    mode='lines+markers+text',
    line=dict(width=5, color='#FF6B6B'),
    marker=dict(size=12, color='#4ECDC4', symbol='circle', line=dict(width=2, color='white')),

    text=labels,   
    texttemplate='<b>%{text}</b>',
    textposition='top center',
    textfont=dict(size=12, color='#292F36'),

    hovertemplate='<b>%{x}</b><br>Received: %{y:.0f}M<extra></extra>'  

# layout clean
fig.update_layout(
    title='<b>Monthly Received Amount</b>',
    title_x=0.5,
    xaxis_title='<b>Months</b>',
    yaxis_title='<b>Received Amount (Millions)</b>',
    font=dict(size=13, family='Arial')
)

fig.show()
```

### Regional Analysis by State for Total Funded Amount

```python
# Group and aggregate
state_funding = df.groupby('address_state')['loan_amount'].sum().sort_values(ascending=True)
state_funding_thousands = (state_funding / 1000).reset_index()

# Plotly horizontal bar chart
fig = px.bar(
    state_funding_thousands,
    x="loan_amount",
    y="address_state",
    orientation='h',
    text=state_funding_thousands["loan_amount"].map(lambda x: f"${x:,.0f}K"),
    color="loan_amount",
    color_continuous_scale="Reds"  
)

# Layout improvements
fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(width=1, color='black'))
)

fig.update_layout(
    title="<b>Total Funded Amount by State (in $ Thousands)</b>",
    title_x=0.5,
    xaxis_title="<b>Funded Amount ($'000)</b>",
    yaxis_title="<b>State</b>",
    font=dict(size=12, family="Arial"),
    plot_bgcolor="white",
    coloraxis_showscale=False, 
    margin=dict(l=80, r=50, t=80, b=50),
    height = 1000
)

fig.show()
```

### Loan Term Analysis by Total Funded Amount

```python
# Step 1: Aggregate data
term_funding = df.groupby('term')['loan_amount'].sum().reset_index()

# Step 2: Convert to millions
term_funding['loan_amount_millions'] = (term_funding['loan_amount'] / 1_000_000).round(1)

# Step 3: Create pie chart
fig = px.pie(
    term_funding,
    names="term",
    values="loan_amount_millions",
    title="<b>Total Funded Amount by Term (in $ Millions)</b>",
    height=550,
    width=900,
    color="term",
    hole=0.5
)

# Step 4: Update layout
fig.update_layout(
    title_font_size=21,
    title_x=0.5,
    font=dict(size=13, family="Arial"),
    annotations=[dict(
        text=f"Funded Amt: {term_funding['loan_amount_millions'].sum():.1f}M",
        x=0.5, y=0.5, font_size=14, showarrow=False,
        bgcolor = 'lightgray',
        bordercolor = 'black',
        borderpad = 4,
        font = dict(size = 34, family = 'Lato Black')
    )],
    legend = dict(
        title = '<b>Terms</b>',
        orientation = 'h',
        xanchor = 'center',
        yanchor = 'bottom',
        x = 0.5,
        y = -0.15
    )
)


# update traces
fig.update_traces(
    textposition = 'outside',
    textfont_size = 16,
    textfont_weight = 'bold',
    texttemplate = '<b>%{label}: %{percent}</b>',
    hovertemplate = '%{label}: $%{value:,}M',
    marker = dict(line = dict(color = 'white', width = 6))
)



fig.show()
```

### Home Ownership by Total Funded Amount

```python
import plotly.express as px

# --- Prepare data ---
home_funding = df.groupby('home_ownership')['loan_amount'].sum().reset_index()
home_funding['loan_amount_millions'] = home_funding['loan_amount'] / 1_000_000

# --- Treemap ---
fig = px.treemap(
    home_funding,
    path=['home_ownership'],
    values='loan_amount_millions',
    color='loan_amount_millions',
    color_continuous_scale='Tealgrn',   # more elegant than Blues
    title='<b>Total Funded Amount by Home Ownership ($ Millions)</b>'
)

# --- Update layout ---
fig.update_layout(
    title_font_size=24,
    title_x=0.5,
    title_font=dict(size=24, family="Arial Black", color="#1E4D78"),
    font=dict(size=14, family="Arial"),
    margin=dict(t=70, l=40, r=40, b=40),
    paper_bgcolor="white",
    showlegend= True
)

# --- Update traces ---
fig.update_traces(
    texttemplate="<b>%{label}</b><br>$%{value:,.1f}M",   # label + value
    textposition="middle center",
    hovertemplate="<b>%{label}</b><br>Funded: $%{value:,.2f}M<br>"
                  "Raw Amount: $%{customdata:,}<extra></extra>",
    customdata=home_funding['loan_amount']  # raw full value
)

fig.show()
```
